### 1. Indentation over Brackets

#### Unlike languages like Java or C++ that use curly brackets {} to organize code blocks, Python uses physical whitespace and indentation. Standard formatting requires 4 spaces per indentation level

```python
# example
score = 99
if score >= 90:
    print("Grade: A")
```

### 2. Variables and Dynamic Typing

#### You do not need to explicitly declare data types when creating a variable; Python determines the type automatically at runtime

```python
# This is a comment. Python ignores it.
age = 25          # Integer
gpa = 3.8          # Floating-point number
name = "Alice"     # String (text)
is_student = True  # Boolean (True/False)
```
