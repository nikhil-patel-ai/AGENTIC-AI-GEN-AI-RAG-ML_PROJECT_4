# Python keywords

## Python keywords are reserved words that have predefined meanings and form the foundational structure of the language. You cannot use them as names for variables, functions, or classes.

### Keywords are reserved words in Python that have a special meaning to the Python interpreter.
### We cannot normally use keywords as variable names, function names, or class names.
