## Tuple creation

```python
tup = ()    # Empty tuple
tup
```

```python
tup1 = (10, 20, 30, 40, 50)    # tuple of integers numbers
tup1
```

```python
tup2 = (3.4, 2.5, 5.3, 10.6)    # tuple of float numbers 
tup2
```

```python
tup3 = ('one', 'two', 'three', 'four')    # tuple of string numbers
tup3
```

```python
tup4 = ('Nikhil', 39,(50,100),(60,75))    # Nested tuples
tup4
```

## Tuple indexing

```python
tup1[0]    # Retreive first element of the tuple
```

```python
tup3[0]
```

```python
tup3[0][0]    # Nested indexing
```

```python
tup3[-1]   # last item of the tuple
```

### Tuple slicing

```python
tup4 =  ('one', 'two', 'three', 'four', 'five', 'six', 'seven', 'eight', 'nine')
tup4
```

```python
tup4[:3]    # return all item from 0th - 3rd index excluding 3rd index
```

```python
tup4[2:5]    # return all item from 2nd index to 5th index excluding 5th index
```

```python
tup4[:3]    # Return first three items
```

```python
tup4[:2]    # Return first two items
```

```python
tup4[-3:]    # last three items 
```

```python
tup4[-2:]    # last two items
```

```python
tup4[-1:]    # last item of tuple
```

```python
tup4[:]    # whole tuple 
```
