# Swap Variable in Python
  #### (a,b = 12,15) After swap we should get ==> (a,b = 15,12)

```python
a = 12
b = 15
print(a,b)
```

```python
a = b
b = a
print(a,b)
```

```python
a,b = b,a
print(a,b)
```

```python
print(a,b)
```

```python
# in above scenario we lost the value 5 
a1 = 4
b1 = 9
print(a1, b1)
```

```python
temp = a1 
a1 = b1
b1 = temp
```

```python
print(a1)
print(b1)
```

```python
a2 = 12
b2 = 11
```

```python
#swap variable formulas
a2 = a2 + b2    # 12 = 12 + 11
b2 = a2 - b2    # 11 = 23 - 11
a2 = a2 - b2    # 12 = 23 - 12
```

```python
print(a2)
print(b2)
```

```python
print(0b101) # 101 is 3 bit 
print(0b110) # 110 also 3bit 
```

```python
#but when we use a2 + b2 then we get 11 that means we will get 4 bit which is 1 bit extra 
print(bin(11))
print(0b1011)
```

```python
#there is other way to work using swap variable also which is XOR because it will not waste extra bit 
a2 = a2 ^ b2
b2 = a2 ^ b2
a2 = a2 ^ b2
```

```python
print(a2)
print(b2)
```

```python
print(a2)
print(b2)
```

```python
a2 , b2 = b2, a2    # We always use this swaping technique
```

```python
print(a2)
print(b2)
```
