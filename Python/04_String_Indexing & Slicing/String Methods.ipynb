# Python String Methods
   #### Here is some most usefull string listed below

## capitalize() Method : 

#### The capitalize() method returns a string where the first character is upper case, and the rest is lower case.

### Syntax

##### string.capitalize()

#### Example :

```python
Cap = ("coding is FUN!")
c = Cap.capitalize()
print(c)
```

## casefold() Method :

 - #### The casefold() method returns a string where all the characters are lower case.

### Syntax

##### string.casefold()

### Example :

```python
case = "Hello, And Welcome To My World!"
x = case.casefold()
print(x)
```

## center() Method :

- #### The center() method will center align the string, using a specified character (space is default) as the fill character.
### Syntax 
 #### string.center(length, character) :
   - ##### length  --> Required. The length of the returned string
   - ##### character  -->  Optional. The character to fill the missing space on each side. Default is " " (space) You can add your own choice
### Example :

```python
cntr = "banana"

x = cntr.center(20, ".")

print(x)
```

 ## 1. Changing the Size (The Case Tools)
#### These buttons change your letters from small to big, or big to small.
 - #### .upper() ➔ Makes everything BIG SHOUTING LETTERS.
 - #### .lower() ➔ Makes everything small quiet letters.
 - #### .title() ➔ Makes the First Letter Of Each Word Big.

```python
cs = "hello world"

print(cs.upper())  # Output: HELLO WORLD
print(cs.lower())  # Output: hello world
print(cs.title())  # Output: Hello World
```

 ## 2. Cleaning Up (The Eraser Tools)
#### Sometimes words have accidental empty spaces at the beginning or end. These tools erase them!
 - #### .strip() ➔ Erases spaces from both ends.
 - #### .lstrip() ➔ Erases spaces from only the left side.
 - #### .rstrip() ➔ Erases spaces from only the right side.

```python
strp = "   cookie   "

print(strp.strip()) # Output: "cookie" (All clean!)
```

## 3. Finding and Replacing (The Magic Wand Tools)
#### These tools help you hunt for words or swap them with new ones.
 - #### .replace("old", "new") ➔ Swaps an old word for a brand new one.
 - #### .find("word") ➔ Tells you the exact number spot where a word starts.

```python
fr = "I love cats!"

print(fr.replace("cats", "dogs")) # Output: I love dogs!
print(fr.find("love"))            # Output: 2 (It starts at spot #2)
```

## 4. Splitting and Gluing (The Scissors and Glue)
#### These are super powerful for breaking sentences apart or putting them back together.
 - #### .split() ➔ Cuts a sentence into a list of separate words.
 - #### .join() ➔ Glues a list of words back together using a special connector.

```python
sg = "apple banana grape"
fruits = sg.split() 
print(fruits) # Output: ['apple', 'banana', 'grape'] (Cut into pieces!)

glue = "-"
print(glue.join(fruits)) # Output: apple-banana-grape (Glued together!)
```

## 5. Asking Questions (The Checker Tools)
#### These buttons don't change the text. They just answer True (Yes) or False (No).
 - #### .startswith("...") ➔ Does the word start with these letters?
 - #### .endswith("...") ➔ Does the word end with these letters?
 - #### .isdigit() ➔ Is this string actually just a pile of numbers?

```python
qa = "photo.png"
print(qa.endswith(".png")) # Output: True
print(qa.isdigit())        # Output: False (It has letters!)
```
