## The in and not in operators

####  1. The in Operator (The "Is It There?" Tool)
- ##### Use in when you want to see if something is inside a group.

```python
toy_box = "lego, doll, car, blocks"
print("car" in toy_box)      # Let's check for a car
# Output: True  (Yes, it's there!)
print("robot" in toy_box)    # Let's check for a robot
# Output: False (No, it's missing!)
```

#### 2. The not in Operator (The "Is It Missing?" Tool)\
- ##### Use not in when you want to make sure something is NOT inside a group.

```python
toy_box = "lego, doll, car, blocks"
print("robot" not in toy_box)    # Let's check if the robot is missing
# Output: True  (Yes, you are right! The robot is NOT in there.)

print("lego" not in toy_box)    # Let's check if the lego is missing
# Output: False (No, you are wrong! The lego IS in there.)
```
