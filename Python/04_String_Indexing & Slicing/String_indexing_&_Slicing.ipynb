+ ## Note :

### 1. Creation and Syntax : 
- #### Quote Flexibility : Use single ('...') or double ("...") quotes to create a string. Both work identically.
- #### Quote Nesting : Use double quotes outside if the string contains a single quote (e.g., "It's okay"), and vice versa.
- #### Multiline Strings : Use triple quotes ("""...""" or '''...''') to span text across multiple lines.
- #### Escape Characters: Use a backslash (\) to insert special characters like newlines (\n) or literal quotes (\').
- #### Raw Strings : Prefix a string with r (e.g., r"C:\Users\name") to treat backslashes as literal characters

### 2. Immutability
- #### Read-Only Data : Python strings are immutable. You cannot modify, add, or delete characters in place.
- #### Creating New Output : Any operation that alters a string (like changing capitalization) creates an entirely new string in memory.

### 3. Indexing and Slicing :
- #### Zero-Based Indexing : The first character is always at index 0, the second is at 1, and so on.
- #### Negative Indexing : Access characters from the end of the string starting at -1 (the last character).
- #### Slicing Syntax : Extract a substring using [start:stop], where the start index is inclusive and the stop index is exclusive.
- #### Out-of-Bounds Error : Accessing a single index that does not exist throws an IndexError. Slicing out of bounds returns an empty string without crashing.

### 4. Basic Operations :
- #### Concatenation : Join strings together using the + operator (e.g., "Hello " + "World").
- #### Replication : Repeat a string multiple times by multiplying it with an integer (e.g., "Hi" * 3 outputs "HiHiHi").
- #### Type Constraints : You cannot concatenate a string directly with a number; you must convert the number first using str(5).
- #### Length : Find the total number of characters (including spaces and punctuation) using the len() function.

## 2. Slicing Types (By Direction)
### Slicing changes its behavior completely based on the sign of your step value.
- #### Positive Step Type (Forward Slicing)Rule : step is a positive integer (or omitted, defaulting to 1).
- #### Direction : Moves left to right.
- #### Requirement : start must be to the left of stop (index start < stop) to return data. If start >= stop, it returns an empty string "".

```python
str2[0: ]    # Start at 0th index and print till end    
```

```python
str2[3:9]    # Start at 3rd index & end with (n-1)th index
```

```python
str2[:4]     # Retreive first four characters of the string 
```

### Negative Step Type (Backward Slicing)
- #### Rule : step is a negative integer (e.g., -1, -2).
- #### Direction : Moves right to left.
- #### Requirement : start must be to the right of stop (index start > stop) to return data. If start <= stop, it returns an empty string "".

```python
str2
```

```python
str2[ :-1]     # Start with -1th index and prints till end
```

```python
str2[-6:]    # Retreive last six characters of the string 
```
