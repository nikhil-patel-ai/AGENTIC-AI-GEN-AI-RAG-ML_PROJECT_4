# 6. Hallow Squre Pattern

```python
for i in range(5):
    for j in range(5):
        if i == 0 or i == 4 or j == 0 or j == 4:
            print('*',end='')
        else:
            print(' ',end='')
    print()
```

# 10. Floyd's Triangle

```python
num = 1 
for i in range(1,6):
    for j in range(1,i+1):
        print(num,end=' ')
        num+=1
    print()
```

# 11. Hallow Right Angle Triangle

```python
for i in range(1,6):
    for j in range(1,i+1):
        if j==1 or j==i or i==5:
            print('*',end=' ')
        else:
            print(' ',end=' ')
    print()
```

# 12. Hallow Pyramid Pattern

```python
for i in range(1,6):
    for j in range(5-i):
        print(' ',end=' ')
    for j in range(2*i-1):
        if j == 0 or j==2*i-2 or i == 5:
            print('*',end=' ')
        else:
            print(' ',end=' ')
    print()
```

# 13. Hallow Diamond Pattern

```python
n = 5
for i in range(1,n+1):
    for j in range(n-i):
        print(' ',end=' ')
    for j in range(2*i-1):
        if j == 0 or j == 2 *i-2:
            print('*',end=' ')
        else:
            print(' ',end=' ')
    print()

for i in range(n-1,0,-1):
    for j in range(n-i):
        print(' ',end=' ')
    for j in range(2*i-1):
        if j == 0 or j == 2 *i-2:
            print('*',end=' ')
        else:
                print(' ',end=' ')
    print()
```

# 14. Hallow Diamond (Number Pattern)

```python
n = 5
for i in range(1,n+1):
    for j in range(n-i):
        print(' ',end=' ')
    for j in range(2*i-1):
        if j == 0 or j == 2 *i-2:
            print(i,end=' ')
        else:
            print(' ',end=' ')
    print()

for i in range(n-1,0,-1):
    for j in range(n-i):
        print(' ',end=' ')
    for j in range(2*i-1):
        if j == 0 or j == 2 *i-2:
            print(i,end=' ')
        else:
                print(' ',end=' ')
    print()
```

# 15. Butterfly Pattern

```python
n = 5
for i in range(1,n+1):
    for j in range(1,i+1):
        print(j,end=' ')
    for j in range(2*(n-i)):
        print(' ',end=' ')
    for j in range(1,i+1):
        print(j,end=' ')
    print()
for i in range(n,0,-1):
    for j in range(1,i+1):
        print(j,end=' ')
    for j in range(2*(n-i)):
        print(' ',end=' ')
    for j in range(1,i+1):
        print(j,end=' ')
    print()
```

# 15.1 Butterfly Pattern (Using *)

```python
n = 5
for i in range(1,n+1):
    for j in range(1,i+1):
        print("*",end=' ')
    for j in range(2*(n-i)):
        print(' ',end=' ')
    for j in range(1,i+1):
        print("*",end=' ')
    print()
for i in range(n,0,-1):
    for j in range(1,i+1):
        print("*",end=' ')
    for j in range(2*(n-i)):
        print(' ',end=' ')
    for j in range(1,i+1):
        print("*",end=' ')
    print()
```

# 16. Hallow Number Pyramid

```python
n = 5
for i in range(1,n+1):
    for j in range(n-i):
        print(' ',end=' ')
        
    for j in range(1,2*i):
        if j == 1 or j == 2*i - 1 or i == n:
            print(i,end=' ')
        else:
            print(' ',end=' ')
    print()
```

# 17. Full Star Pyramid

```python
n = 5 
for i in range(1,n+1):
    for j in range(n-i):
        print(' ',end=' ')
    for j in range(2*i-1):
        print('*',end=' ')
    print()
```

# 18. Inverted Full Star Pyramid

```python
n = 5
for i in range(n,0,-1):
    for j in range(n-i):
        print(' ',end=' ')
    for j in range(2*i-1):
        print('*',end=' ')
    print()
```

# 19. Left Aligned Pyramid Pattern

```python
# left aligned pyramid pattern with *
n = 5
for i in range(1,n+1):
    for j in range(i):
        print('*',end=' ')
    print()
```

```python
# left aligned pyramid pattern with numbers
n = 5
for i in range(1,n+1):
    for j in range(1,i+1):
        print(j,end=' ')
    print()
```

# 20. Right Aligned Pyramid Pattern

```python
# right aligned pyramid pattern with number
n = 5 
for i in range(1,n+1):
    for j in range(n-i):
        print(' ',end=' ')
    for j in range(1,i+1):
        print(j,end=' ')
    print()
```

```python
# right aligned pyramid pattern with using '*' .
n = 5 
for i in range(1,n+1):
    for j in range(n-i):
        print(' ',end=' ')
    for j in range(1,i+1):
        print("*",end=' ')
    print()
```
