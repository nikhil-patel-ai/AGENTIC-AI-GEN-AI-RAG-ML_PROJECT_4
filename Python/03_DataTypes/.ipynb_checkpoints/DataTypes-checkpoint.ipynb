# Data Types

#### -Numeric --> int, float, complex --->  Whole numbers, decimals, and imaginary numbers.   ---->   42, 3.14, 2+3j
#### -Text --> str(string)  --->  Character sequences (strings) enclosed in quotes.   ---->   "Hello, Python!"
#### -Sequence --> list, tuple, range  --->  Ordered collections of items.   ---->   [1, 2, 3], (4, 5), range(5)
#### -Mapping --> dict  --->  Unordered key-value pairs.   ---->   {"name": "Alice", "age": 25}
#### -Set --> set, frozenset  --->  Unordered collections of unique elements.   ---->   {1, 2, 3}
#### -Boolean --> bool  --->  Logical states representing truth values.   ---->   True, False
#### -None --> NoneType  --->  Represents the absence of a value.   ---->   None

### 1. Numeric Types

#### int :
####     Positive or negative whole numbers with unlimited precision.

```python
# example
i = 20
type(i)
```

```python
i1 = 40
type(i1)
```

#### float : 
####       Real numbers containing a decimal point.

```python
# example for float(contains decimal value) data type
f = 54.23
type(f)
```

```python
f1 = 20.23
type(f)
```

### 2. Text Type :
###          Used to manage textual data.

#### str : 
#### Strings are immutable sequences of Unicode characters wrapped in single, double, or triple quotes.
##### Example :

```python
str1 = 'Wellcome' # this is single quote string also use for single line string
print(str1)
type(str1)
```

```python
s2 = " This is doublequotes string "    # uesed for single line string
print(s2)
type(s2)
```

```python
s3 = ''' This also a string
this is used for write multiline string ''' 
type(s3)
print(s3)
type(s3)
```

### 3. Sequence Types :
#### Used to store multiple data items in an ordered arrangement.
##### Example :

#### list : 
   #### Mutable (changeable) sequences that can store mixed data types.
##### Example :

```python
l =  [1, "apple", True]
print(type(l))
print(l)
```

```python
l1 =  (3, "Mango", False)   # we can not use round brackets in list, when we use round brackets type is changed list to tuple
print(l1)
print(type(l1))
```

```python
l2 =  {1, "apple", True} # also we can not  use curly brackets in list, when we use curly brackets type is changed list to set
type(l2)
```

```python
l3 =  <1, "apple", True>  # also we can not use angle brackets
type(l2)
```

```python
l4 =  [1, "apple", True]
```

#### tuple : 
#### Immutable (unchangeable) sequences primarily used to protect data from accidental alterations.
##### Example :

```python
t  = (10.5, 20.3)
print(t)
type(t)
```

#### same rules here like list rules

### range : 
#### Represents an immutable sequence of numbers, commonly utilized for looping specific frequencies. 
##### Example :

```python
r = range(1, 10)
print(r)
print(type(r))
```

## 4. Mapping Type :
### Used to store data sets that have relationships.
#### dict : 
#### Dictionaries map unique keys to specific values. They are highly efficient for looking up data.
##### Example : 

```python
d = {"id": 101, "role": "admin"}
type(d)
```

## 5. Set Types
### Used for membership testing and deduplication.
### set : 
#### An unordered collection containing unique items.
##### Example :

```python
s = {101, 102, 103}
print(s)
print(type(s))
```

### frozenset: 
#### An immutable variation of a standard set.
##### Example :

```python
fr = frozenset([1, 2, 3])
print(fr)
print(type(fr))
```

```python
fr1 = frozenset([1, 2, 4])
print(fr1)
print(type(fr1))
```
