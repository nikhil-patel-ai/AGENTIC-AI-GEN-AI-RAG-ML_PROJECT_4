# Patterns printing in Python

```python
print(' # # # # ')
print(' # # # # ')
print(' # # # # ')
print(' # # # # ')
```

```python
for i in range(1,5):
    i  = i + 1
    print('# # # #')
```

```python
for i in range(1,5):
    i  = i <= 5
    print('# # # #')
```

```python
for i in range(4):
    for j in range(i+1):
        print('#', end = " ")
    print()
```

```python
for i in range(4):
    for j in range(4-i):
        print('#', end = " ")
    print()
```
