#### Q1. What is the input() function in Python used for?
#### Ans- The input() function in Python used for taking input from user.

#### Q2. How can you accept an integer as input from the user using input()?
#### Ans - using int keyword code is below

```python
i = int(input("Please enter number : " ))
print("User entered : ",i)
```

#### Q3. How do you accept a float input from the user?
#### Ans - using float keyword, code given beloww :

```python
f = float(input("Please enter float number : " ))
print("Float is : ",f)
```
